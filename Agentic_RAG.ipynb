```python
from langchain_community.document_loaders import PyPDFLoader

# PDF 檔案路徑
pdf_path = "/content/勞動基準法qa百問百答第二版.pdf"

# 載入 PDF
loader = PyPDFLoader(pdf_path)

docs_list = loader.load()

print(f"成功載入 {len(docs_list)} 頁")
```

```python
!pip install PyMuPDF
import fitz
from langchain_core.documents import Document

# 開啟 PDF
doc = fitz.open(pdf_path)

page_documents = []

for page_index, page in enumerate(doc):

    text = page.get_text().strip()

    # 空白頁不加入
    if not text:
        continue

    page_documents.append(
        Document(
            page_content=text,
            metadata={
                "source": "勞動基準法QA百問百答第二版",
                "page": page_index + 1
            }
        )
    )

print("PDF 總頁數：", len(doc))
print("成功建立 Page Documents：", len(page_documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separators=["\n\n", "\n", "。", "，", " ", ""]
)

chunks = text_splitter.split_documents(page_documents)

print("Page Documents：", len(page_documents))
print("切分後 Chunks：", len(chunks))
```

```python
import chromadb
from langchain_chroma import Chroma

# Get or create a Chroma client. Since no persist_directory is specified,
# this implicitly connects to the in-memory Chroma instance.
chroma_client = chromadb.Client()

collection_name = "labor_law_rag_final"

try:
    # Attempt to delete the collection if it already exists from a previous run
    chroma_client.delete_collection(name=collection_name)
    print(f"Successfully deleted existing Chroma collection: '{collection_name}'.")
except Exception as e:
    # This exception will occur if the collection doesn't exist, which is fine.
    print(f"Collection '{collection_name}' not found or could not be deleted (might not exist yet): {e}")

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name=collection_name,
    client=chroma_client # Pass the explicit client to ensure it's used
)

print("Vector DB 建立完成")
print("Chunk 數量：", len(chunks))
```

```python
from FlagEmbedding import FlagReranker

reranker = FlagReranker(
    'BAAI/bge-reranker-base',
    use_fp16=True
)
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 10}
)
```

```python
def rerank_documents(user_question, docs, top_k=3):
    """
    使用 BGE Reranker 對 Vector DB 找到的文件重新排序
    """

    # 建立 Query-Document 配對
    pairs = [
        [user_question, doc.page_content]
        for doc in docs
    ]

    # 計算相關性分數
    scores = reranker.compute_score(
        pairs,
        normalize=True
    )

    # 將文件與分數配對
    ranked_docs = sorted(
        zip(docs, scores),
        key=lambda x: x[1],
        reverse=True
    )

    # 取出 Top K
    top_docs = [
        doc
        for doc, score in ranked_docs[:top_k]
    ]

    return ranked_docs, top_docs
```

```python
question = "勞工每天最多可以加班幾小時？"

# 第一階段：Vector Retrieval Top 10
retrieved_docs = vectorstore.as_retriever(
    search_kwargs={"k": 10}
).invoke(question)

# 第二階段：BGE Reranker Top 3
ranked_docs, top_docs = rerank_documents(
    question,
    retrieved_docs,
    top_k=3
)

for i, (doc, score) in enumerate(ranked_docs[:3], 1):

    print(f"\n===== Rerank Top {i} =====")
    print("Page:", doc.metadata.get("page"))
    print("Score:", score)
    print(doc.page_content[:500])
```

```python
def ask_question(
    user_question,
    top_k_retrieval=10,
    top_k_rerank=3
):

    # Vector Search
    docs = vectorstore.as_retriever(
        search_kwargs={
            "k": top_k_retrieval
        }
    ).invoke(user_question)

    # Reranker
    ranked_docs, top_docs = rerank_documents(
        user_question,
        docs,
        top_k=top_k_rerank
    )

    # Context
    context = "\n\n".join(
        doc.page_content
        for doc in top_docs
    )

    prompt = f"""
你是一個專門回答台灣《勞動基準法》相關問題的智慧問答助手。

請根據下方提供的參考資料回答使用者問題。

回答規則：
1. 只能根據提供的參考資料回答。
2. 不要自行捏造法規內容。
3. 如果資料不足，請明確說明。
4. 請使用繁體中文。
5. 回答要清楚、簡潔、容易理解。

【參考資料】
{context}

【使用者問題】
{user_question}

【回答】
"""

    response = llm.invoke(prompt)

    print("【回答】")
    print(response.content)

    print("\n【參考頁面】")

    for doc in top_docs:
        print(
            "PDF Page:",
            doc.metadata.get("page")
        )

    return response
```

```python
def evaluate_reranker_hit_at_3(evaluation_dataset, vectorstore):
    results = []

    for item in evaluation_dataset:
        question = item["question"]
        ground_truth_pages = item["ground_truth_pages"]

        # Baseline：直接 Vector Search Top 3
        docs = vectorstore.as_retriever(
            search_kwargs={"k": 3}
        ).invoke(question)

        retrieved_pages = [
            doc.metadata.get("page") for doc in docs
        ]

        # 只要 Top 3 任一頁命中 Ground Truth，就算成功
        hit = any(
            page in ground_truth_pages
            for page in retrieved_pages
        )

        results.append({
            "id": item["id"],
            "question": question,
            "type": item["type"],
            "ground_truth_pages": ground_truth_pages,
            "retrieved_pages": retrieved_pages,
            "hit@3": hit
        })

    return results
```

```python
reranker_results = evaluate_reranker_hit_at_3(
    evaluation_dataset,
    vectorstore
)

for result in reranker_results:
    print(
        result["id"],
        "| Ground Truth:", result["ground_truth_pages"],
        "| Retrieved:", result["retrieved_pages"],
        "| Hit:", result["hit@3"]
    )
```

```python
reranker_hit_at_3 = sum(
    result["hit@3"] for result in reranker_results
) / len(reranker_results)

print(f"Reranker Hit@3：{reranker_hit_at_3:.2%}")
```

```python
def calculate_mrr(results):
    reciprocal_ranks = []

    for result in results:
        ground_truth_pages = result["ground_truth_pages"]
        retrieved_pages = result["retrieved_pages"]

        rr = 0

        for rank, page in enumerate(retrieved_pages, start=1):
            if page in ground_truth_pages:
                rr = 1 / rank
                break

        reciprocal_ranks.append(rr)

    return sum(reciprocal_ranks) / len(reciprocal_ranks)


baseline_mrr = calculate_mrr(reranker_results)

print(f"Reranker MRR：{baseline_mrr:.4f}")
```

```python
def rag_answer(user_question, top_k_retrieval=10, top_k_rerank=3):
    if not user_question.strip():
        return "請先輸入問題。", ""

    # 1. Vector Search
    docs = vectorstore.as_retriever(
        search_kwargs={"k": top_k_retrieval}
    ).invoke(user_question)

    # 2. Reranker
    ranked_docs, top_docs = rerank_documents(
        user_question,
        docs,
        top_k=top_k_rerank
    )

    # 3. 組合 Context
    context = "\n\n".join(
        doc.page_content for doc in top_docs
    )

    # 4. Prompt
    prompt = f"""
你是一個專門回答台灣《勞動基準法》相關問題的智慧問答助手。

請根據下方提供的參考資料回答使用者問題。

回答規則：
1. 只能根據提供的參考資料回答。
2. 不要自行捏造法規內容。
3. 如果資料不足，請明確說明。
4. 請使用繁體中文。
5. 回答要清楚、簡潔、容易理解。

【參考資料】
{context}

【使用者問題】
{user_question}

【回答】
"""

    # 5. LLM
    response = llm.invoke(prompt)

    # 6. 整理參考頁碼
    pages = []
    for doc in top_docs:
        page = doc.metadata.get("page")
        if page is not None and page not in pages:
            pages.append(page)

    page_text = "、".join(str(page) for page in pages)

    return response.content, f"PDF 參考頁面：{page_text}"
```
